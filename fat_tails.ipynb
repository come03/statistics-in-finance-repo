```python
xmin, xmax = 1e-2,1
ymax = 0.1

alpha = -2.1

x = np.linspace(xmin,xmax, 1000)


plt.plot(x, x**alpha, label = 'not-scaled')

c = ymax/xmin**alpha
plt.plot(x,c*x**alpha, label = "scaled")

plt.yscale('log')
plt.xscale('log')

plt.xlabel('X')
plt.ylabel('f(X)')
plt.legend()
plt.show()
```

```python
def plot_powerlaw(alpha, xmin, xmax, ymax,  n_steps = 1000):
    c = ymax/xmin**alpha
    x = np.linspace(xmin, xmax, n_steps)
    return x,c*x**alpha
```

```python
plt.plot(*plot_powerlaw(alpha = -2, xmin = 0.01, xmax = 0.2, ymax = 0.1))

plt.xlabel('X')
plt.ylabel('f(X)')
plt.show()
```

```python
T = returns.shape[0]
dt=1000
steps = 15 

dates = returns.index

estimators = []
for t in np.arange(0,T-dt,steps):
    
    r = returns.iloc[t : t+dt]
    
    nu,_,_ = st.t.fit(r)
    alpha = pl.Fit(abs(r)).alpha
    
    estimators.append((dates[t+dt],nu,alpha))
```

```python
estimators = pd.DataFrame(estimators, columns = ['dates','T-Student Fit','Powerlaw Fit'])

estimators.set_index('dates').plot()
plt.ylabel('Exponent')
```
